```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
```

```python
student_data = {
    "Student_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Amara", "Dev", "Jyoti", "Rahul", "Priya", "Kabir"],
    "Score": [85, 42, 78, 91, 35, 62]
}
df = pd.DataFrame(student_data)

# Define grades based on scores
def get_grade(score):
    if score >= 90:
        return 'A'
    elif score >= 70:
        return 'B'
    elif score >= 50:
        return 'C'
    else:
        return 'D'

df['Grade'] = df['Score'].apply(get_grade)
display(df)
```

Now, let's prepare the data for KNN classification. We'll use 'Score' as the feature to predict 'Grade'.

```python
# Features (X) and target (y)
X = df[['Score']]
y = df['Grade']

# Split data into training and testing sets (stratify removed due to small dataset size and single-member classes)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Initialize and train the KNN classifier
# We'll use n_neighbors=2 as an example since we have very few samples per class
knn = KNeighborsClassifier(n_neighbors=2)
knn.fit(X_train, y_train)

# Make predictions
y_pred = knn.predict(X_test)

print("Predictions:", y_pred)
print("Actual Grades:", y_test.values)
```

Let's evaluate the performance of our KNN classifier.

```python
from sklearn.metrics import accuracy_score, classification_report

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred, zero_division=0)

print(f"Accuracy: {accuracy:.2f}")
print("Classification Report:\n", report)
```
